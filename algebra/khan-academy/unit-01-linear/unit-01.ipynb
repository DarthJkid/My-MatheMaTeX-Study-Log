```python
from sympy import symbols, Eq, solve, solve_univariate_inequality, S, oo, Rational
from sympy import latex, pretty_print as pp
import numpy as np
import matplotlib.pyplot as plt

x = symbols('x')
```

## 1  Solving Linear Equations

Theorem (Unique Solution): $ax + b = 0,\ a \neq 0$ has the unique solution $x = -b/a$.

```python
# Exercise 1: solve 3x - 7 = 2
eq1 = Eq(3*x - 7, 2)
sol1 = solve(eq1, x)
print(f"3x - 7 = 2  =>  x = {sol1}")
```

```python
# General form: solve ax + b = 0
from sympy import symbols as sym
a, b = sym('a b', real=True, nonzero=True)
general_eq = Eq(a*x + b, 0)
print("General solution:", solve(general_eq, x))
```

## 2  Solving Linear Inequalities

Remember: dividing by a **negative** number reverses the inequality.

```python
# Exercise 2: solve -2x + 4 >= 10
from sympy import GreaterThan
ineq = -2*x + 4 >= 10
sol_ineq = solve_univariate_inequality(ineq, x, relational=False)
print(f"-2x + 4 >= 10  =>  x in {sol_ineq}")
```

## 3  Visualization: solution sets on a number line

```python
fig, ax = plt.subplots(figsize=(8, 1.5))

ax.axhline(0, color='black', linewidth=1)
ax.set_xlim(-6, 2)
ax.set_ylim(-0.5, 0.5)

# shade solution set (-inf, -3]
ax.fill_betweenx([-0.15, 0.15], -6, -3, color='steelblue', alpha=0.4,
                 label=r'$x \leq -3$')
ax.plot(-3, 0, 'o', color='steelblue', markersize=10)  # closed dot at -3

ax.set_yticks([])
ax.set_xlabel('x')
ax.set_title(r'Solution set of $-2x + 4 \geq 10$')
ax.legend(loc='upper right')

plt.tight_layout()
plt.show()
```
